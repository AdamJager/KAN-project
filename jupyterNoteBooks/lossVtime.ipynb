```python
import os, sys
import numpy as np
import torch
import matplotlib.pyplot as plt
module_path = os.path.abspath(os.path.join('..'))
sys.path.append(os.path.join(module_path, "KAN_variants"))
from ChebyKANLayer import ChebyKAN
from efficientKAN import KAN
from fastkan import FastKAN

module_path = os.path.abspath(os.path.join('..'))
sys.path.append(os.path.join(module_path, "code"))
from myKANODE import KANODE
```

```python
def pred_prey_deriv(X, t, alpha, beta, delta, gamma):
    x=X[0]
    y=X[1]
    dxdt = alpha*x-beta*x*y
    dydt = delta*x*y-gamma*y
    dXdt=[dxdt, dydt]
    return dXdt

# alpha, beta, delta, gamma
params = [1.5, 1, 2, 1]
X0 = [0.3, 1.5]
```

```python
def vanDerPoll(X, t, epsilon):
    x = X[0]
    y = X[1]
    dxdt = y
    dydt = epsilon*(1-x*x)*y - x
    return [dxdt, dydt]
vdpX0 = [0, 1.5]
vdpParams = [0.2]
```

```python
model = ChebyKAN([2, 10, 2], 4)
t0 = 20
t1 = 500
kanODE = KANODE(model, vanDerPoll, vdpX0, vdpParams, t1, t0, 0.008)
```

```python
saveDirectory = os.path.join(module_path, "savedModels", "vdp3")
```

```python
kanODE.loadModel(saveDirectory, "secondBatchLearn1439")
```

```python
_, solution = kanODE.test("rk4")
loss = torch.square(solution[kanODE.trainingSamples:,0, :]-kanODE.baseSolution[kanODE.trainingSamples:, :]).detach().cpu()
```

```python
kanODE.plotODE("ode and solution", solution[:, 0, :])
kanODE.plotPhaseSpace("Phase Space", solution[:, 0, :])
```

```python
time = torch.linspace(t0, t1, len(loss[:, 0]))
```

```python
plt.plot(time, loss[:, 0])
plt.plot(time, loss[:, 1])
```
