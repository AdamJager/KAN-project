```python
import os, sys
import numpy as np
import matplotlib.pyplot as plt
module_path = os.path.abspath(os.path.join('..'))
sys.path.append(module_path+"\\KAN_variants")
from ChebyKANLayer import ChebyKAN

module_path = os.path.abspath(os.path.join('..'))
sys.path.append(module_path+"\\code")
from myKANODE import KANODE
```

```python
def pred_prey_deriv(X, t, alpha, beta, delta, gamma):
    x=X[0]
    y=X[1]
    dxdt = alpha*x-beta*x*y
    dydt = delta*x*y-gamma*y
    dXdt=[dxdt, dydt]
    return dXdt
X0 = [0.1, 1.5]
Params = [1.5, 1, 2, 1]
```

```python
ICs = [[ii/10, jj/10] for ii in range(1, 11, 2) for jj in range(2, 24, 4)]
```

```python
saveDirectory = os.path.join(module_path, "savedModels", "initialConditionSweep")
```

```python
trainedPointsX = np.array([])
trainedPointsY = np.array([])
for index, X0 in enumerate(ICs):
    try:
        model = ChebyKAN([2, 10, 2], 4)
        kanODE = KANODE(model, pred_prey_deriv, X0, Params, 20, 6, 0.007)
        kanODE.train(-1, maxEpochs = 2000, evalThreshold = 10**-4, saveDirectory=saveDirectory, saveName=f"ics-{X0}_epoch-")
        trainedPointsX = np.append(trainedPointsX, X0[0])
        trainedPointsY = np.append(trainedPointsY, X0[1])
    except:
        continue
```
